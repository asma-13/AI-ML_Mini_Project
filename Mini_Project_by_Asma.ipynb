<h3>importing all the nessary libraries,classes and functions. and reading the dataset with the help of pandas library</h3>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
dataset=pd.read_csv('C:/Users/Dell/Downloads/archive/data.csv')
```

<h3>checking if there are any null values in the dataset</h3>

```python
dataset.isna()
```

<h3>Dropping and clearing all the uneccesary null values from the dataset</h3>

```python
dataset.dropna()
dataset
```

<h3>Making the DataFrame of data in order to perform effienct access and control of data</h3>

```python
df=pd.DataFrame(dataset)
df
```

<h3>Calculating the Mean, Mode, Normal  Distribution and Standard Deviation of Height Column</h3>

```python
mean_of_height=np.mean(df['Hight'])
mode_of_height= stats.mode(df['Hight'])
nd_height=np.random.normal(df['Hight'])
std_height=np.std(df['Hight'])
print('The Mean Hight is: ',mean_of_height)
print('The Mode of Hight is: ',mode_of_height)
print('The Standard Deviation of Hight is: ',std_height)
plt.hist(nd_height)
```

<h2>Interpretetion Of Hight Results:</h2>
<ul>
    <li>
        <h4>Mean Height: </h4> This indicates the average height in dataset is 173cm. Which represents the central tendency of data.
    </li>
    <li>
        <h4>Mode of Height: </h4> The most common height in dataset is 172cm, meaning a significant number of individuals share this height.
    </li>
    <li>
        <h4>Standard Deviation of Height: </h4> A higher standard deviation (13.52 cm) suggests that heights are spread out, with notable variation among individuals.
    </li>
</ul>
<h4>Insight:</h4>The average height is 173.40 cm, with a mode of 172 cm. The standard deviation is 13.52 cm, indicating a moderate spread in height variations among individuals. The height data appears to follow a normal distribution, suggesting that most individuals are clustered around the mean with a few taller or shorter individuals.
<h4>Recommendation: </h4>Since height follows a normal distribution, this can be useful in designing products like clothing, door heights, or sports equipment, ensuring they fit the majority population.
